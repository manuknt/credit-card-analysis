# 02 — Limpeza e Preparação dos Dados

## Objetivo

Nesta etapa, serão realizadas verificações de qualidade dos dados, análise de variáveis redundantes e seleção das características que poderão ser utilizadas na modelagem preditiva.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/processed/dados_limpos.csv")

print("Dados carregados com sucesso!")
print("Dimensões:", df.shape)

display(df.head())
```

```python
print("Colunas:")
print(df.columns.tolist())

print("\nTipos de dados:")
print(df.dtypes)

print("\nValores ausentes:")
print(df.isnull().sum())

print("\nRegistros duplicados:", df.duplicated().sum())
```

```python
print("Distribuição de aprovação (%):")

display(
    (df["aprovado"].value_counts(normalize=True) * 100).round(2)
)
```

```python
print("Renda média por resultado:")
display(df.groupby("aprovado")["renda"].mean())

print("Dívida média por resultado:")
display(df.groupby("aprovado")["divida"].mean())

print("Taxa de aprovação por inadimplência anterior:")
display(df.groupby("inadimplencia_anterior")["aprovado"].mean())

print("Pontuação de crédito média por resultado:")
display(df.groupby("aprovado")["pontuacao_de_credito"].mean())
```

## Seleção das Variáveis para Modelagem

A seleção das variáveis considerou a qualidade dos dados, a relevância para o problema e os cuidados necessários na interpretação das informações.

Foram adotadas as seguintes decisões:

- Manter variáveis relacionadas ao perfil financeiro e ao histórico de crédito.
- Excluir variáveis sensíveis, como gênero e etnia, do modelo inicial.
- Excluir código postal, por ser um identificador geográfico.
- Excluir inadimplência anterior, pois seus códigos não possuem documentação suficiente para interpretação segura.
- Excluir empregado(a), pois essa variável apresenta redundância com a pontuação de crédito.
- Evitar variáveis cuja interpretação não esteja suficientemente documentada.

A seleção representa uma primeira abordagem exploratória e não um modelo definitivo para decisões reais de concessão de crédito.

```python
colunas_modelo = [
    "idade",
    "divida",
    "anos_empregado",
    "pontuacao_de_credito",
    "renda",
    "cliente_do_banco",
    "aprovado"
]

df_modelo = df[colunas_modelo].copy()

print("Dimensões da base preparada:", df_modelo.shape)

display(df_modelo.head())
```

```python
print("Valores ausentes:")
print(df_modelo.isnull().sum())

print("\nRegistros duplicados:")
print(df_modelo.duplicated().sum())

print("\nDistribuição da variável-alvo:")
display(
    (df_modelo["aprovado"].value_counts(normalize=True) * 100).round(2)
)
```

```python
df_modelo.to_csv(
    "../data/processed/dados_modelagem.csv",
    index=False
)

print("Base preparada salva com sucesso!")
```

## Conclusão da Preparação dos Dados

A base foi carregada e analisada, apresentando 672 registros e 16 variáveis originais.

Não foram identificados valores ausentes ou registros duplicados na base original.

Após a seleção das variáveis, foi criada uma base com 6 variáveis preditoras e 1 variável-alvo, totalizando 672 registros e 7 colunas.

A base preparada foi salva em `data/processed/dados_modelagem.csv` e utilizada na etapa de modelagem preditiva.

As escolhas de variáveis consideraram a relevância para o problema, a redundância entre informações e os cuidados necessários com variáveis sensíveis e sem documentação suficiente.
