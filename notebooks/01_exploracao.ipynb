```python
import numpy as np

# Criar novas colunas sem alterar as originais
df["Renda_log"] = np.log1p(df["Renda"])
df["Pontuacao_Credito_log"] = np.log1p(
    df["Pontuação de Crédito"]
)

# Comparar a assimetria antes e depois
comparacao = pd.DataFrame({
    "Variável": [
        "Renda",
        "Renda transformada",
        "Pontuação de Crédito",
        "Pontuação transformada"
    ],
    "Assimetria": [
        df["Renda"].skew(),
        df["Renda_log"].skew(),
        df["Pontuação de Crédito"].skew(),
        df["Pontuacao_Credito_log"].skew()
    ]
})

display(comparacao.round(3))
```

```python
# Criar faixas de renda
df["Faixa_Renda"] = pd.cut(
    df["Renda"],
    bins=[-1, 0, 500, 1500, 5000, float("inf")],
    labels=[
        "Zero",
        "Até 500",
        "501 a 1500",
        "1501 a 5000",
        "Acima de 5000"
    ]
)

# Analisar aprovação por faixa de renda
analise_renda = df.groupby(
    "Faixa_Renda",
    observed=False
)["Aprovado"].agg(
    quantidade="count",
    aprovados="sum",
    taxa_aprovacao="mean"
)

analise_renda["taxa_aprovacao"] *= 100

display(analise_renda.round(2))
```

### Análise da taxa de aprovação por faixa de renda

A análise da aprovação por faixas de renda revelou uma associação positiva entre renda e aprovação de crédito.

Os clientes com renda entre 1 e 500 apresentaram a menor taxa de aprovação, de 30,08%. Já na faixa entre 501 e 1.500, a taxa aumentou para 72,97%.

Nas faixas superiores, a aprovação continuou aumentando, chegando a 79,17% entre 1.501 e 5.000 e a 92,31% para rendas acima de 5.000.

Os clientes com renda igual a zero apresentaram uma taxa de aprovação de 37,85%, superior à faixa de renda entre 1 e 500.

**Conclusão:** a renda apresenta uma associação importante com a aprovação de crédito neste conjunto de dados. Entretanto, os resultados são descritivos e não demonstram causalidade. Também é necessário considerar o tamanho reduzido dos grupos com renda mais elevada.

Essa variável será mantida para as próximas etapas da análise e da modelagem.

```python
# Criar faixas de pontuação de crédito

df["Faixa_Pontuacao"] = pd.cut(
    df["Pontuação de Crédito"],
    bins=[-1, 0, 5, 10, 20, float("inf")],
    labels=[
        "Zero",
        "1 a 5",
        "6 a 10",
        "11 a 20",
        "Acima de 20"
    ]
)

# Analisar aprovação por faixa

analise_pontuacao = df.groupby(
    "Faixa_Pontuacao",
    observed=False
)["Aprovado"].agg(
    quantidade="count",
    aprovados="sum",
    taxa_aprovacao="mean"
)

analise_pontuacao["taxa_aprovacao"] *= 100

display(analise_pontuacao.round(2))
```

### Análise da taxa de aprovação por faixa de pontuação de crédito

A análise revelou uma associação expressiva entre a pontuação de crédito e a aprovação.

Os clientes com pontuação igual a zero representam 388 registros (57,74% da base) e apresentam uma taxa de aprovação de 24,48%.

Na faixa de 1 a 5 pontos, a taxa de aprovação sobe para 57,65%. Entre 6 e 10 pontos, chega a 87,69%. Na faixa de 11 a 20 pontos, alcança 93,48%.

Os três clientes com pontuação acima de 20 foram aprovados, mas o tamanho reduzido desse grupo não permite generalizações confiáveis.

**Conclusão:** a pontuação de crédito apresenta uma associação forte com a aprovação neste conjunto de dados. A variável será mantida para a modelagem. Os valores zero serão investigados antes de qualquer decisão sobre seu tratamento, pois ainda não sabemos se representam ausência de informação ou pontuação efetivamente igual a zero.
